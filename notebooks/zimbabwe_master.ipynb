# Zimbabwe: the master table (v1)

`master` = one row per field, with everything joined: labels, season indices, Presto, MOSAIKS.
The months are not in `master`. They are in the view `monthly` (cell 3).

```python
import duckdb
import pandas as pd

pd.set_option("display.max_columns", 50)
con = duckdb.connect("../data/cropyield.duckdb", read_only=True)   # run from notebooks/
sql = lambda query: con.execute(query).df()

# 1. Zimbabwe, model fields only (include_primary): zimbabwe_2024 + zimbabwe_lacuna_2022_2023
zim = sql("SELECT * FROM master WHERE country = 'zimbabwe' AND include_primary")
print(zim.shape)
zim.head()
```

```python
# 2. Other things: pick columns, one cohort, or any other WHERE
lacuna = sql("""
    SELECT obs_id, cohort, yield_kg_ha, months_with_data, ndvi_season_max, gcvi_season_max
    FROM master
    WHERE cohort = 'zimbabwe_lacuna_2022_2023' AND include_primary
""")
print(lacuna.shape)
lacuna.head()
```

```python
# 3. The months: view `monthly` = one row per field and month (bands, indices, clear or not)
months = sql("""
    SELECT obs_id, cohort, month, coverage_state, b04, b08, ndvi, gcvi, yield_kg_ha
    FROM monthly
    WHERE obs_id IN (SELECT obs_id FROM master WHERE country = 'zimbabwe' AND include_primary)
    ORDER BY obs_id, month
""")
print(months.shape)
months.head(9)    # the 9 months of the first field
```

```python
# 4. From master to months: master has no months, so take its obs_id and look them up in `monthly`
m = sql("SELECT obs_id, cohort, month, ndvi FROM monthly").merge(zim[["obs_id"]], on="obs_id")
ndvi = m[m.cohort == "zimbabwe_2024"].pivot(index="obs_id", columns="month", values="ndvi")
zim_2024 = zim.merge(ndvi, on="obs_id")      # one master row + one NDVI column per month
zim_2024[["obs_id", "yield_kg_ha", *ndvi.columns]].head()
```
