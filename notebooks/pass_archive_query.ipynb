```python
from pathlib import Path

import duckdb

REPO = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
DB = REPO / "data" / "passarchive_v1" / "release" / "passarchive_v1.duckdb"      # change this to open another file
con = duckdb.connect(str(DB), read_only=True)


def sql(query):
    """Run one SQL query on the database and return a pandas table."""
    return con.execute(query).df()


# what is inside: the tables with their row counts, then the views
sql("""select table_name as name, 'table' as kind, estimated_size as rows from duckdb_tables() where not internal
       union all select view_name, 'view', null from duckdb_views() where not internal order by 2, 1""")
```
